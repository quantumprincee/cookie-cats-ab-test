# Cookie Cats: сдвигать ворота с 30 на 40?

Данные: [Mobile Games A/B Testing](https://www.kaggle.com/datasets/yufengsui/mobile-games-ab-testing), 90 189 игроков.

Контроль — ворота на 30 уровне, тест — на 40. Первичная метрика — вернулся ли игрок на 7-й день. День 1 и число раундов — сторожевые. Катим `gate_40` только если день 7 не хуже контроля.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from math import erf, sqrt

sns.set_theme(style="whitegrid")
df = pd.read_csv("../data/cookie_cats.csv")
df["retention_1"] = df["retention_1"].astype(bool)
df["retention_7"] = df["retention_7"].astype(bool)
df.head()
```

## Осмотр

Одна строка — один игрок. Дубли `userid` значили бы, что человек мог попасть в обе группы.

```python
print("строк", len(df))
print("уникальных userid", df["userid"].nunique())
print("дублей", df["userid"].duplicated().sum())
print(df.isna().sum())
print(df["version"].value_counts())
```

## SRM

Ждали примерно 50/50. На большой выборке p < 0.05 ещё не повод выбрасывать тест. Промышленный стоп чаще 0.001.

```python
n_30 = (df["version"] == "gate_30").sum()
n_40 = (df["version"] == "gate_40").sum()
expected = (n_30 + n_40) / 2
chi2 = (n_30 - expected) ** 2 / expected + (n_40 - expected) ** 2 / expected
srm_p = 2 * (1 - 0.5 * (1 + erf(sqrt(chi2) / sqrt(2))))

print("gate_30", n_30, "gate_40", n_40)
print("доля gate_30", round(n_30 / len(df), 4))
print("chi2", round(chi2, 3), "p", round(srm_p, 4))
print("стоп на 0.001:", srm_p < 0.001)
```

## Удержание

Доля, разница в процентных пунктах и 95% интервал. p-value — z-тест для двух долей.

```python
rows = []
for metric in ["retention_1", "retention_7"]:
    control = df.loc[df["version"] == "gate_30", metric]
    treatment = df.loc[df["version"] == "gate_40", metric]
    p_30 = control.mean()
    p_40 = treatment.mean()
    diff = p_40 - p_30
    se = np.sqrt(p_30 * (1 - p_30) / len(control) + p_40 * (1 - p_40) / len(treatment))
    p_pool = (control.sum() + treatment.sum()) / (len(control) + len(treatment))
    se_pool = np.sqrt(p_pool * (1 - p_pool) * (1 / len(control) + 1 / len(treatment)))
    z = diff / se_pool
    pvalue = 2 * (1 - 0.5 * (1 + erf(abs(z) / sqrt(2))))
    rows.append({
        "metric": metric,
        "gate_30": round(p_30, 4),
        "gate_40": round(p_40, 4),
        "diff_pp": round(diff * 100, 2),
        "ci_low": round((diff - 1.96 * se) * 100, 2),
        "ci_high": round((diff + 1.96 * se) * 100, 2),
        "pvalue": round(pvalue, 4),
    })

result = pd.DataFrame(rows)
result
```

```python
plot_df = df.groupby("version")[["retention_1", "retention_7"]].mean().reset_index()
plot_df = plot_df.melt(id_vars="version", var_name="metric", value_name="rate")
plot_df["rate"] = plot_df["rate"] * 100
plot_df["metric"] = plot_df["metric"].map({"retention_1": "день 1", "retention_7": "день 7"})

plt.figure(figsize=(7, 4))
sns.barplot(data=plot_df, x="version", y="rate", hue="metric", order=["gate_30", "gate_40"])
plt.ylabel("Доля игроков, %")
plt.xlabel("")
plt.title("Удержание не выросло, когда ворота сдвинули на 40")
plt.tight_layout()
plt.savefig("../figures/retention_by_gate.png", dpi=140)
plt.show()
```

День 7 ниже на 0.82 п.п., интервал от −1.33 до −0.31, p = 0.0016. День 1 тоже ниже, но интервал захватывает ноль. Правило «не хуже контроля» не выполнено.

## Раунды

Число раундов снято уже после теста, это не признак для CUPED. У одного игрока контроля 49 854 раунда, среднее из-за него врёт.

```python
outlier = df.loc[df["sum_gamerounds"].idxmax()]
print(outlier["userid"], outlier["version"], outlier["sum_gamerounds"])
print("медиана")
print(df.groupby("version")["sum_gamerounds"].median())
print("среднее")
print(df.groupby("version")["sum_gamerounds"].mean().round(2))
print("среднее без выброса")
print(df.loc[df["sum_gamerounds"] < 3000].groupby("version")["sum_gamerounds"].mean().round(2))
print("доля нулевых раундов")
print(df.assign(zero=df["sum_gamerounds"] == 0).groupby("version")["zero"].mean().round(4))
```

```python
rounds = df.loc[df["sum_gamerounds"] < 3000].copy()
plt.figure(figsize=(7, 4))
sns.boxplot(
    data=rounds,
    x="version",
    y="sum_gamerounds",
    order=["gate_30", "gate_40"],
    showfliers=False,
)
plt.ylabel("Раунды за 14 дней")
plt.xlabel("")
plt.title("Медиана раундов почти не различается")
plt.tight_layout()
plt.savefig("../figures/rounds_boxplot.png", dpi=140)
plt.show()
```

## Срез по 30 раундам — только посмотреть

Дошёл игрок до 30 раундов или нет, зависит от ворот. Резать по этому полю и принимать решение нельзя.

```python
df["hit_30"] = df["sum_gamerounds"] >= 30
df.groupby(["version", "hit_30"])["retention_7"].agg(["mean", "size"]).round(4)
```
